```python
#------- EXAMPLE OUTER EXOSPHERE PLOTS -----------------------------------------
import exospy.exospy as ep
import numpy as np
#import matplotlib.pyplot as plt

# Create a Class exosgrid
myexosgrid = ep.exosgrid()

# Initial values for geometric structure 
myexosgrid.rmin   = 3
myexosgrid.rmax   = 8
myexosgrid.rstep  = 0.25
myexosgrid.pmin   = 0
myexosgrid.pmax   = 360
myexosgrid.pstep  = 10
myexosgrid.tmin   = 0
myexosgrid.tmax   = 180
myexosgrid.tstep  = 5

# Calculate internal values
myexosgrid.calculate_intvals()
```

```python
model = "bailey_2008"

H1 = ep.generate3DHmodel(model, myexosgrid)

print("type:", type(H1))
print("shape:", H1.shape)           # (numvoxels, 1)
print("min/max:", H1.min(), H1.max())
print("finite:", np.all(np.isfinite(H1)))
```

```python
toPlot, H = ep.draw3DHmodel("bailey_2008", myexosgrid, "equatorial", 0.0, True)
```
